```python
from ultralytics import YOLO
model = YOLO("yolov8s.pt")
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    name="drishti_model",
    device=0  # use GPU
)
```

```python
# Validating the model
metrics = model.val()
print(metrics)
```

```python
results = model.predict("/content/test.png", show=False)
```

```python
import matplotlib.pyplot as plt
import cv2

threshold = 0 # Set the confidence threshold

# Plot the results and display them in the notebook
for r in results:
    # Filter boxes based on confidence threshold
    high_conf_mask = r.boxes.conf > threshold
    filtered_boxes = r.boxes[high_conf_mask]

    # Temporarily replace the original boxes with the filtered ones for plotting
    original_boxes = r.boxes
    r.boxes = filtered_boxes

    res_plotted = r.plot() # Now r.plot() will use the filtered boxes

    # Restore original boxes (optional, but good practice if `results` is used elsewhere)
    r.boxes = original_boxes

    plt.figure(figsize=(10, 10))
    plt.imshow(cv2.cvtColor(res_plotted, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.title(f"Detections with confidence > {threshold}")
    plt.show()
```
